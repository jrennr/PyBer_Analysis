```python
from pathlib import Path
import pandas as pd

student_df = pd.read_csv(Path('new_full_student_data.csv'))
print(student_df.isna().sum())
student_df = student_df.dropna().drop_duplicates().copy()
student_df['grade'] = student_df['grade'].astype(str).str.replace('th', '', regex=False).astype('int64')
student_df.head()
```

```python
print(student_df.dtypes)
print(student_df.describe())
print('Mean math score:', student_df['math_score'].mean())
min_reading_score = student_df['reading_score'].min()
print('Minimum reading score:', min_reading_score)
school_info = student_df.iloc[:19514]
school_info
```
